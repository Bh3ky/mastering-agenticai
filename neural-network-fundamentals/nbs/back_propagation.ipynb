# Training a Neural Network with PyTorch

### Lab: Linear Regression, Loss Functions, Backpropagation & Gradient Descent

In this lab, we will train a simple neural network to learn the relationship:

$$

y=2x+1
$$

We will provide input-output examples and allow the model to learn the underlying relationship by adjusting its parameters.


```python
import torch
```

```python
# sample dataset
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
Y = torch.tensor([[3.0], [5.0], [7.0], [9.0]])
```

```python
print("Input shape:", X.shape)
print("Target shape:", Y.shape)
```

```python
print("Inputs:")
print(X)
```

```python
print("Targets:")
print(Y)
```

```python
torch.manual_seed(42) # for reproducibility
model = torch.nn.Linear(1, 1) # one input and one output feature
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01) # Stochastic Gradient Descent optimizer
```

```python
print("Initial weight:", model.weight.item())
print("Initial bias:", model.bias.item())
```

```python
predictions = model(X)
```

```python
print("Predictions:")
print(predictions)
print("Predictions shape:", predictions.shape)
```

```python
loss_fn = torch.nn.MSELoss()
loss = loss_fn(predictions, Y)
```

```python
print("Loss:", loss.item())
```

```python
loss.backward()
```

```python
print("Weight gradient:", model.weight.grad)
print("Bias gradient:", model.bias.grad)
```

```python
optimizer.step()
```

```python
print("Updated weight:", model.weight.item())
print("Updated bias:", model.bias.item())
```

```python
new_predictions = model(X)
new_loss = loss_fn(new_predictions, Y)
print("New loss:", new_loss.item())
```

```python
losses = []
```

```python
for epoch in range(5000):
    optimizer.zero_grad()

    predictions = model(X)
    loss = loss_fn(predictions, Y)

    loss.backward()
    optimizer.step()

    losses.append(loss.item())

    if (epoch + 1) % 500 == 0:
        print(f"Epoch {epoch + 1}: loss = {loss.item():.6f}")

print("Learned weight:", model.weight.item())
print("Learned bias:", model.bias.item())

final_predictions = model(X)
final_loss = loss_fn(final_predictions, Y)

print("Final predictions:")
print(final_predictions.detach())

print("Final loss:", final_loss.item())
```
